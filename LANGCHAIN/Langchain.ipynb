```python
import os

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model(
    'openai/gpt-oss-120b',
     model_provider='groq',
     temperature = 0.7)

response = model.invoke("What is your Name?")

print(response.content)
```

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# Load variables from .env file
load_dotenv()

# Check API key
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError("GOOGLE_API_KEY not found. Check your .env file.")

# Initialize Gemini model
model = init_chat_model(
    "gemini-3.5-flash-lite",
    model_provider="google_genai",
    temperature=0.2,
)

# Ask question
response = model.invoke(
    "Tell me about students studying at QIS College of Engineering and Technology in Ongole."
)

# Print response
print(response.content)
```

```python
import os

from langchain.chat_models import init_chat_model

os.environ["MISTRAL_API_KEY"] = os.getenv("MISTRAL_API_KEY")

model = init_chat_model(
    "mistral-large-latest",
     model_provider='mistralai',
     temperature=0.2,)

response = model.invoke("Tell me about  Animal movie in telugu")

print(response.content)
```

```python
model.invoke("Give me about Agentic Ai about 50 words")
```

```python
for chunk in model.stream("Give me about Agentic AI about 50 words"):
    print(chunk.text,end = " ", flush=True)
```

```python
from IPython.display import display

responses = model.batch([
    "How will students get Generative AI jobs?",
    "What qualifications are required for a Generative AI job?",
    "Which place is considered the happiest place in the world?",
])

for i, result in enumerate(responses, start=1):
    display(f"### Answer {i}")
    display(result.content)
    
```

```python
responses = model.batch([
    "what is Generative AI?",
    "How do we reach a HR's in Linked Easily?",
    "Which is the best peaceful place in the ongole?"],
    config={
        'max_occurency' : 5,
    })
for response in responses:
    print(response.text)
```
